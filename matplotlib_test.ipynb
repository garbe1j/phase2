### Plots are created using only pandas and matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("weather_data.csv")
df
```

```python
plt.figure(figsize=(6, 3))
plt.plot(df["day"], df["temp_high"], marker="o", label="High")
plt.plot(df["day"], df["temp_low"], marker="o", label="Low")
plt.title("Daily High and Low Temperatures")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.bar(df["day"], df["precip_mm"], color="skyblue")
plt.title("Daily Precipitation")
plt.xlabel("Day")
plt.ylabel("Precipitation (mm)")
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.scatter(df["wind_kmh"], df["temp_high"], color="red")
plt.title("Wind Speed vs. Temperature")
plt.xlabel("Wind Speed (km/h)")
plt.ylabel("High Temperature (°C)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.hist(df["temp_high"], bins=5, color="green")
plt.title("Distribution of High Temperatures")
plt.xlabel("High Temperature (°C)")
plt.ylabel("Frequency")
plt.show()
```
